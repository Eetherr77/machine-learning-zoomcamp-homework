```python
import numpy as np

# 1. Select cars from Asia
asia_cars = df[df["origin"] == "Asia"]

# 2–3. Select columns and first 7 rows
X = asia_cars[["vehicle_weight", "model_year"]].head(7).to_numpy()

# 4–5. Calculate X transpose multiplied by X
XTX = X.T @ X

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Calculate w
w = XTX_inv @ X.T @ y

# 9. Sum all elements of w
print(w.sum())
```
